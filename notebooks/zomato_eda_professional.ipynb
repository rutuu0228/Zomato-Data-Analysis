### 🎯 Objective
Turn raw restaurant data into clear, decision-oriented insights using Python.

## 📌 Business Questions

1. Which restaurant type receives the most customer orders?
2. How many votes does each restaurant type receive?
3. What rating do most restaurants receive?
4. What is the average spending per order?
5. Which ordering mode receives higher ratings?
6. Which restaurant type receives more offline orders?

## 🧰 1. Import Libraries

We use Pandas for data manipulation and Matplotlib/Seaborn for visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 📂 2. Load the Dataset

```python
DATA_PATH = '../Zomato data .csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print(f'Dataset shape: {df.shape}')
print(f'Rows: {df.shape[0]:,}')
print(f'Columns: {df.shape[1]:,}')
df.info()
```

## 🧹 3. Data Cleaning & Preparation

The rating column contains values such as `4.1/5`. We convert it to a numeric value so that it can be analyzed statistically. We also normalize column names for easier use in Python.

```python
df = df.copy()
df.columns = (
    df.columns.str.strip()
             .str.lower()
             .str.replace(' ', '_')
             .str.replace(r'[^a-z0-9_()]', '', regex=True)
)

df['rate'] = (
    df['rate'].astype(str)
      .str.replace('/5', '', regex=False)
      .replace({'nan': np.nan, '': np.nan})
      .astype(float)
)

df.head()
```

```python
print('Missing values:')
display(df.isna().sum())

print(f'Duplicate rows: {df.duplicated().sum()}')
```

## 📊 4. Exploratory Analysis

### 4.1 🍴 Most Popular Restaurant Type

This identifies the restaurant category with the largest number of listings in the dataset.

```python
restaurant_type_counts = df['listed_in(type)'].value_counts()
restaurant_type_counts
```

```python
ax = restaurant_type_counts.plot(kind='bar', title='Restaurant Listings by Type')
ax.set_xlabel('Restaurant Type')
ax.set_ylabel('Number of Restaurants')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 4.2 👍 Customer Votes by Restaurant Type

```python
votes_by_type = df.groupby('listed_in(type)')['votes'].sum().sort_values(ascending=False)
display(votes_by_type)

sns.barplot(x=votes_by_type.index, y=votes_by_type.values)
plt.title('Total Customer Votes by Restaurant Type')
plt.xlabel('Restaurant Type')
plt.ylabel('Total Votes')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 4.3 ⭐ Distribution of Restaurant Ratings

```python
rating_distribution = df['rate'].value_counts().sort_index()
display(rating_distribution)

sns.histplot(df['rate'].dropna(), bins=10, discrete=True)
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Rating out of 5')
plt.ylabel('Number of Restaurants')
plt.tight_layout()
plt.show()
```

### 4.4 💰 Average Cost for Two People

```python
cost_col = 'approx_cost(for_two_people)'
average_cost = df[cost_col].mean()
print(f'Average cost for two people: ₹{average_cost:,.2f}')
```

### 4.5 🛵 Online vs Offline Ordering

```python
online_rating = df.groupby('online_order')['rate'].mean().sort_values(ascending=False)
display(online_rating)

sns.barplot(x=online_rating.index, y=online_rating.values)
plt.title('Average Rating: Online vs Offline Ordering')
plt.xlabel('Online Order')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.tight_layout()
plt.show()
```

### 4.6 🏪 Restaurant Types with More Offline Orders

```python
offline_orders = (
    df.loc[df['online_order'].eq('No')]
      .groupby('listed_in(type)')
      .size()
      .sort_values(ascending=False)
)

display(offline_orders)

offline_orders.plot(kind='bar', title='Offline Orders by Restaurant Type')
plt.xlabel('Restaurant Type')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
